# 🧮 Special Matrices in Code — NumPy & PyTorch

What you'll do:
- Build identity, diagonal, and triangular matrices with **NumPy**, and compute inverses/determinants with `np.linalg`
- Do the **exact same operations** with **PyTorch tensors**
- Run a matrix inverse through **autograd**

Run each cell top to bottom.

## Part 1: NumPy
### Cell 1 — Setup

Same A = [[2,1],[1,3]] from "Special Matrices Practice" (det(A) = 5).

```python
import numpy as np

A = np.array([[2, 1], [1, 3]])
print(f"A =\n{A}")
```

### Cell 2 — Identity matrix

```python
I2 = np.eye(2)
print(f"I2 =\n{I2}")
print(f"I2 @ A =\n{I2 @ A}")   # always A

B = np.array([[2, 3, 1], [1, 4, 0]])   # 2x3, non-square
I3 = np.eye(3)
print(f"B @ I3 =\n{B @ I3}")   # I3 (3x3) on the right, matches B's column count
```

### Cell 3 — Diagonal matrix

```python
D = np.diag([2, 3, 4])
E = np.diag([5, 1, 2])
print(f"D =\n{D}")
print(f"D + E =\n{D + E}")
print(f"D @ E =\n{D @ E}")   # off-diagonal terms are automatically 0
```

### Cell 4 — Diagonal power, inverse, determinant

```python
D2 = np.diag([2, -1])
print(f"D2 ** 10 =\n{np.linalg.matrix_power(D2, 10)}")   # [[1024, 0], [0, 1]]

D3 = np.diag([2, -4, 5])
print(f"inv(D3) =\n{np.linalg.inv(D3)}")   # reciprocal of each diagonal entry

D4 = np.diag([2, 3, 4])
print(f"det(D4) = {np.linalg.det(D4)}")   # product of the diagonal entries
```

### Cell 5 — Triangular matrix (extract from any matrix)

```python
M = np.array([[2, 3, 1], [5, 1, 5], [3, 4, 4]])
print(f"tril(M) =\n{np.tril(M)}")   # keeps on/below the diagonal, zeros the rest
print(f"triu(M) =\n{np.triu(M)}")   # keeps on/above the diagonal, zeros the rest
```

### Cell 6 — Inverse of A, verified

```python
A_inv = np.linalg.inv(A)
print(f"det(A) = {np.linalg.det(A)}")
print(f"A_inv =\n{A_inv}")
print(f"A @ A_inv =\n{A @ A_inv}")   # should be I
```

### Cell 7 — Try it yourself
✏️ Change the numbers below and re-run.

```python
# ✏️ Try changing this matrix (keep it square with a nonzero determinant)
M2 = np.array([[3.0, 1.0], [2.0, 2.0]])
print(f"det(M2) = {np.linalg.det(M2)}")
print(f"inv(M2) =\n{np.linalg.inv(M2)}")
print(f"M2 @ inv(M2) =\n{M2 @ np.linalg.inv(M2)}")
```

## Part 2: PyTorch — the same operations
### Cell 8 — Identity, diagonal, inverse, determinant

```python
import torch

At = torch.tensor(A, dtype=torch.float32)

print(f"eye(2) =\n{torch.eye(2)}")
print(f"diag =\n{torch.diag(torch.tensor([2., 3., 4.]))}")
print(f"det(At) = {torch.linalg.det(At)}")
print(f"inv(At) =\n{torch.linalg.inv(At)}")
```

### Cell 9 — Inverse through autograd

Same gradient idea as the other notebooks — this time the gradient flows through `torch.linalg.inv`.

```python
At2 = torch.tensor(A, dtype=torch.float32, requires_grad=True)

A_inv_t = torch.linalg.inv(At2)
loss = A_inv_t.sum()
loss.backward()

print(f"A_inv (via PyTorch) =\n{A_inv_t.detach()}")
print(f"gradient w.r.t. A =\n{At2.grad}")
```

### Cell 10 — Try it yourself
✏️ Change the numbers and see how the gradient changes.

```python
# ✏️ Try changing this matrix
M3 = torch.tensor([[4.0, 2.0], [1.0, 3.0]], requires_grad=True)

M3_inv = torch.linalg.inv(M3)
loss = M3_inv.sum()
loss.backward()

print(f"M3      =\n{M3.detach()}")
print(f"inv(M3) =\n{M3_inv.detach()}")
print(f"gradient =\n{M3.grad}")
```

## 🎉 Done!

| Concept | NumPy | PyTorch |
|---|---|---|
| Identity | `np.eye(n)` | `torch.eye(n)` |
| Diagonal | `np.diag([...])` | `torch.diag(tensor)` |
| Determinant | `np.linalg.det(A)` | `torch.linalg.det(A)` |
| Inverse | `np.linalg.inv(A)` | `torch.linalg.inv(A)` |
| Matrix power | `np.linalg.matrix_power(D, n)` | `torch.linalg.matrix_power(D, n)` |
| Lower/upper triangular | `np.tril(M)` / `np.triu(M)` | `torch.tril(M)` / `torch.triu(M)` |

**Everything from "Special Matrices" and "Special Matrices Practice" is a single
function call in real code.** The formulas you worked through by hand — swap,
negate, divide for the inverse; entry-by-entry for the diagonal — are exactly what
these functions do internally, just optimized and generalized to any size.
