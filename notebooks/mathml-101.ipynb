# 🧮 Vector Calculations in Code — NumPy & PyTorch

- Represent vectors as **NumPy arrays** and compute addition, scalar multiplication, dot product, and magnitude
- Do the **exact same operations** with **PyTorch tensors**

## Part 1: NumPy
### Cell 1 — Setup

```python
import numpy as np

# Same customer vector from the course: 3 purchases, 7 visits, 2 support tickets
a = np.array([3, 7, 2])
b = np.array([1, 2, 1])

print(f"a = {a}")
print(f"b = {b}")
```

### Cell 2 — Addition & scalar multiplication

```python
# Addition: combine two vectors, position by position
print(f"a + b = {a + b}")

# Scalar multiplication: scale every position by one number
print(f"2 * a = {2 * a}")
```

### Cell 3 — Dot product & magnitude

Same numbers used in the course's similarity comparison.

```python
a2 = np.array([3, 7])
b2 = np.array([4, 6])

# Dot product: multiply matching positions, then sum
dot = np.dot(a2, b2)   # equivalent: a2 @ b2
print(f"a2 . b2 = {dot}")

# Magnitude: length of the vector
mag = np.linalg.norm(a2)
print(f"||a2||  = {mag:.4f}")
```

### Cell 4 — Try it yourself
✏️ Change the numbers below and re-run.

```python
# ✏️ Try changing these vectors
v1 = np.array([2, 5])
v2 = np.array([3, 1])

print(f"v1 + v2 = {v1 + v2}")
print(f"3 * v1  = {3 * v1}")
print(f"v1 . v2 = {np.dot(v1, v2)}")
print(f"||v1||  = {np.linalg.norm(v1):.4f}")
```

## Part 2: PyTorch — the same operations
### Cell 5 — Setup

```python
import torch

# Exactly the same numbers, now as PyTorch tensors
a = torch.tensor([3., 7., 2.])
b = torch.tensor([1., 2., 1.])

print(f"a = {a}")
print(f"b = {b}")
```

### Cell 6 — Same four operations, PyTorch syntax

Notice the syntax is almost identical to NumPy.

```python
print(f"a + b = {a + b}")
print(f"2 * a = {2 * a}")

a2 = torch.tensor([3., 7.])
b2 = torch.tensor([4., 6.])
print(f"a2 dot b2 = {torch.dot(a2, b2)}")
print(f"||a2||    = {torch.norm(a2):.4f}")
```

### Cell 7 — Autograd: why AI actually uses PyTorch

NumPy can compute everything above, but it has no idea what a *gradient* is.
PyTorch tracks every operation on a tensor and can compute derivatives
automatically — this is what makes training a model possible.

```python
# requires_grad=True tells PyTorch to track this tensor
w = torch.tensor([3.0, 7.0], requires_grad=True)

# A tiny "loss" — sum of squares, same shape as a magnitude² calculation
loss = (w ** 2).sum()
print(f"loss = {loss.item()}")

# Ask PyTorch to compute d(loss)/d(w) automatically
loss.backward()
print(f"gradient = {w.grad}")   # derivative of x^2 is 2x -> [6.0, 14.0]
```

### Cell 8 — Try it yourself
✏️ Change the numbers and see how the gradient changes.

```python
# ✏️ Try changing this vector
w = torch.tensor([2.0, 5.0], requires_grad=True)

loss = (w ** 2).sum()
loss.backward()

print(f"w        = {w.detach()}")
print(f"loss     = {loss.item()}")
print(f"gradient = {w.grad}")
```

## Part 3: NumPy — Matrices
### Cell 9 — Setup

Same dataset X (3 customers × 3 features) and weights W (3 features × 2 outputs) from the course.

```python
X = np.array([[3, 7, 2], [6, 14, 4], [8, 1, 9]])   # 3 customers x 3 features
W = np.array([[0.5, -0.2], [0.1, 0.8], [-0.3, 0.4]])  # weights: 3 features x 2 outputs

print(f"X =\n{X}")
print(f"W =\n{W}")
```

### Cell 10 — Addition & scalar multiplication

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 1], [2, 3]])

print(f"A + B =\n{A + B}")
print(f"2 * A =\n{2 * A}")
```

### Cell 11 — Transpose

```python
print(f"X =\n{X}")
print(f"X.T =\n{X.T}")   # rows become columns
```

### Cell 12 — Matrix multiplication

X (3×3) times W (3×2) — the same X · W from the course.

```python
result = X @ W   # equivalent: np.matmul(X, W)
print(f"X @ W =\n{result}")
```

### Cell 13 — Try it yourself
✏️ Change the numbers below and re-run.

```python
# ✏️ Try changing these matrices
M1 = np.array([[1, 2], [3, 4]])
M2 = np.array([[5, 6], [7, 8]])

print(f"M1 @ M2 =\n{M1 @ M2}")
```

---
## Part 4: PyTorch — the same matrix operations
### Cell 14 — Setup + matrix multiplication

```python
Xt = torch.tensor(X, dtype=torch.float32)
Wt = torch.tensor(W, dtype=torch.float32)

print(f"Xt @ Wt =\n{Xt @ Wt}")   # same result as X @ W above
```

### Cell 15 — Matrix multiplication through autograd

Same gradient idea as the vector notebook, now with W as a matrix instead of a single vector.

```python
Wt2 = torch.tensor(W, dtype=torch.float32, requires_grad=True)
Xt2 = torch.tensor(X, dtype=torch.float32)

out = Xt2 @ Wt2
loss = out.sum()
loss.backward()

print(f"gradient w.r.t. W =\n{Wt2.grad}")
```

## 🎉 Done!

| Concept | NumPy | PyTorch |
|---|---|---|
| Vector | `np.array([...])` | `torch.tensor([...])` |
| Matrix | `np.array([[...], [...]])` | `torch.tensor([[...], [...]])` |
| Addition | `a + b` | `a + b` |
| Scalar multiply | `2 * a` | `2 * a` |
| Dot product | `np.dot(a, b)` | `torch.dot(a, b)` |
| Magnitude | `np.linalg.norm(a)` | `torch.norm(a)` |
| Transpose | `X.T` | `X.T` |
| Matrix multiply | `X @ W` | `X @ W` |
| Gradient | ❌ not supported | ✅ `.backward()` |

**This is the actual foundation every AI model is built on** — the same
vector and matrix operations from this chapter, computed thousands of
times per second during training, with PyTorch automatically tracking
the gradients needed to learn.
