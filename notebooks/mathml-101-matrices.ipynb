# 🧮 Matrix Calculations in Code — NumPy & PyTorch

What you'll do:
- Represent matrices as **NumPy arrays** and compute addition, scalar multiplication, transpose, and matrix multiplication
- Do the **exact same operations** with **PyTorch tensors**
- Run a matrix multiplication through **autograd** — the same gradient machinery from the vector notebook, now on matrices

## Part 1: NumPy
### Cell 1 — Setup

Same dataset X (3 customers × 3 features) and weights W (3 features × 2 outputs) from the course.

```python
import numpy as np

X = np.array([[3, 7, 2], [6, 14, 4], [8, 1, 9]])      # 3 customers x 3 features
W = np.array([[0.5, -0.2], [0.1, 0.8], [-0.3, 0.4]])  # weights: 3 features x 2 outputs

print(f"X =\n{X}")
print(f"W =\n{W}")
```

### Cell 2 — Addition & scalar multiplication

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 1], [2, 3]])

print(f"A =\n{A}")
print(f"B =\n{B}")
print(f"A + B =\n{A + B}")
print(f"2 * A =\n{2 * A}")
```

### Cell 3 — Transpose

```python
print(f"X =\n{X}")
print(f"X.T =\n{X.T}")   # rows become columns
```

### Cell 4 — Matrix multiplication

X (3×3) times W (3×2) — the same X · W from the course.

```python
result = X @ W   # equivalent: np.matmul(X, W)
print(f"X @ W =\n{result}")
```

### Cell 5 — Try it yourself
✏️ Change the numbers below and re-run.

```python
# ✏️ Try changing these matrices
M1 = np.array([[1, 2], [3, 4]])
M2 = np.array([[5, 6], [7, 8]])

print(f"M1 =\n{M1}")
print(f"M2 =\n{M2}")
print(f"M1 @ M2 =\n{M1 @ M2}")
```

## Part 2: PyTorch — the same operations
### Cell 6 — Setup + matrix multiplication

```python
import torch

Xt = torch.tensor(X, dtype=torch.float32)
Wt = torch.tensor(W, dtype=torch.float32)

print(f"Xt @ Wt =\n{Xt @ Wt}")   # same result as X @ W above
```

### Cell 7 — Matrix multiplication through autograd

Same gradient idea as the vector notebook, now with W as a matrix instead of a single vector.

```python
Wt2 = torch.tensor(W, dtype=torch.float32, requires_grad=True)
Xt2 = torch.tensor(X, dtype=torch.float32)

out = Xt2 @ Wt2
loss = out.sum()
loss.backward()

print(f"gradient w.r.t. W =\n{Wt2.grad}")
```

### Cell 8 — Try it yourself
✏️ Change the numbers and see how the gradient changes.

```python
# ✏️ Try changing these matrices
Wt3 = torch.tensor([[1.0, 2.0], [3.0, 4.0]], requires_grad=True)
Xt3 = torch.tensor([[1.0, 0.0], [0.0, 1.0]])

out = Xt3 @ Wt3
loss = out.sum()
loss.backward()

print(f"Wt3      =\n{Wt3.detach()}")
print(f"Xt3      =\n{Xt3}")
print(f"out      =\n{out.detach()}")
print(f"loss     = {loss.item()}")
print(f"gradient =\n{Wt3.grad}")
```

## 🎉 Done!

| Concept | NumPy | PyTorch |
|---|---|---|
| Matrix | `np.array([[...], [...]])` | `torch.tensor([[...], [...]])` |
| Addition | `A + B` | `A + B` |
| Scalar multiply | `2 * A` | `2 * A` |
| Transpose | `X.T` | `X.T` |
| Matrix multiply | `X @ W` | `X @ W` |
| Gradient | ❌ not supported | ✅ `.backward()` |

**This is exactly what a neural network layer does** — data `X`
multiplied by weights `W`, with PyTorch automatically tracking the
gradients needed to update those weights during training.
